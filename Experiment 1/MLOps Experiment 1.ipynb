```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score,classification_report
import seaborn as sns
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"
columns = ['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'target']
```

```python
df = pd.read_csv(url,header=None,names=columns,na_values='?')
df.head(7)
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df["target"] = (df["target"] > 0).astype(int)
```

```python
x = df.drop(columns=['target'])
x
```

```python
y = df.target
y
```

```python
params_svm = {
    "kernel" : "linear",
    "C" : 1.0,
    "gamma" : "scale"
}
```

```python
params_rfc = {
    "n_estimators" : 100,
    "max_depth" : 3,
    "random_state" : 43
}
```

```python
def model_training(x,y,classifier):
    x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=43,stratify=y)
    scaler = StandardScaler()
    x_train = scaler.fit_transform(x_train)
    x_test = scaler.transform(x_test)
    model = classifier.fit(x_train,y_train)
    y_pred = model.predict(x_test)
    accuracy = accuracy_score(y_test,y_pred)
    c_report = classification_report(y_test,y_pred)
    return accuracy,c_report,y_test,y_pred
```

```python
svm = SVC(**params_svm)
```

```python
accuracy_svm, classification_report_svm,y_test_svm,y_pred_svm = model_training(x,y,svm)
print(f"The accuracy of SVM model is: {accuracy_svm*100:.2f}")
```

```python
print(classification_report_svm)
```

```python
rfc = RandomForestClassifier(**params_rfc)
```

```python
accuracy_rfc, classification_report_rfc,y_test_rfc,y_pred_rfc = model_training(x,y,rfc)
print(f"The accuracy of Random Forest Classifier model is: {accuracy_rfc*100:.2f}")
```

```python
print(classification_report_rfc)
```

```python
report_svm = classification_report(y_test_svm,y_pred_svm,output_dict=True)
report_svm
```

```python
report_rfc = classification_report(y_test_rfc,y_pred_rfc,output_dict=True)
report_rfc
```

```python
svm_classifier = {
    'model_name' : 'Support Vector Machine',
    'params' : params_svm,
    'metrics' : report_svm,
    'model' : svm
}
```

```python
random_forest = {
    'model_name' : 'Random Forest Classifier',
    'params': params_rfc,
    'metrics' : report_rfc,
    'model' : rfc
}
```

```python
models = [svm_classifier,random_forest]
```

```python
import mlflow
```

```python
mlflow.set_experiment("MLOps Experiment 1")
```

```python
mlflow.set_tracking_uri(uri="http://localhost:5000")
```

```python
for i,element in enumerate(models):
    model_name = element['model_name']
    params = element['params']
    metrics = element['metrics']
    model = element['model']
    with mlflow.start_run(run_name=model_name):
        mlflow.log_params(params)
        mlflow.log_metrics({
            'accuracy' : metrics['accuracy'],
            'recall_no_heart_disease' : metrics['0']['recall'],
            'recall_heart_disease' : metrics['1']['recall'],
            'precision_no_heart_disease' : metrics['0']['precision'],
            'precision_heart_disease' : metrics['1']['precision'],
            'f1_score_macro' : metrics['macro avg']['f1-score']
        })
        mlflow.sklearn.log_model(model,model_name)
```
